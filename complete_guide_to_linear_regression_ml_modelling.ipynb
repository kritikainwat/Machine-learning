## **📈 Linear Regression The OLS-School Way**
### **⚡ Coefficients, Intercept, and Business Use Cases**  
Linear Regression is a statistical method for modeling the relationship between a **dependent variable $(Y)$** and one or more **independent variables $(X)$**.  

It is **simple yet powerful** and widely applied in business domains for **predicting outcomes, optimizing pricing, improving customer retention, and enhancing operational efficiency**. 🚀  

---

## **1️⃣ Linear Regression Formula Breakdown**  
Linear regression is used for **predicting numerical values** based on the relationship between **independent variables (predictors)** and a **dependent variable (target)**.  

### **📌 Simple Linear Regression Formula**  
For a single independent variable:  

$$
Y = \beta_0 + \beta_1 X + \epsilon
$$  

### **📌 Multiple Linear Regression Formula**  
For multiple independent variables:  

$$
Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + ... + \beta_n X_n + \epsilon
$$  

### **📌 Explanation of Terms**  
🔹 **$Y$ (Target Variable):** The value being predicted.  
🔹 **$X_1, X_2, ..., X_n$ (Predictors):** Features that influence $Y$.  
🔹 **$\beta_0$ (Intercept):** The baseline $Y$ value when all $X_i$ = 0.  
🔹 **$\beta_1, \beta_2, ..., \beta_n$ (Slopes):** The effect of each $X$ on $Y$.  
🔹 **$\epsilon$ (Error Term):** Accounts for variations not captured by $X$.  

💡 *Example:* In real estate price prediction:  
- **$X_1$ (Square Footage)**
- **$X_2$ (Number of Bedrooms)**
- **$X_3$ (Distance to City Center)**  

If $\beta_1 = 5000$, then **each additional square foot increases price by $5000**.  

---


## **1️⃣.2️⃣ Finding Slope ($\beta_1$) and Intercept ($\beta_0$)**  
### **📌 Formula for Slope ($\beta_1$)**  
The slope tells how much **$Y$** changes when **$X$** increases by **1 unit**.  

$$
b_1 = \frac{n \sum (X_i Y) - \sum X_i \sum Y}{n \sum X_i^2 - (\sum X_i)^2}
$$  

Or:  

$$
b_1 = \frac{\sum (X_i - \bar{X})(Y - \bar{Y})}{\sum (X_i - \bar{X})^2}
$$  

Where:  
🔹 **$X_i$** = Each individual $X$ value  
🔹 **$Y$** = Each individual $Y$ value  
🔹 **$\bar{X}$** = Mean of $X$ values  
🔹 **$\bar{Y}$** = Mean of $Y$ values  

📌 **Numerator** → Measures how $X$ and $Y$ move together (covariance).  
📌 **Denominator** → Measures how much $X$ spreads (variance).  

---

### **📌 Formula for Intercept ($\beta_0$)**  
Once we find $\beta_1$, we compute $\beta_0$:  

$$
b_0 = \frac{\sum Y - b_1 \sum X_1 - b_2 \sum X_2}{n}
$$  

Or:  

$$
b_0 = \bar{Y} - b_1 \bar{X_1} - b_2 \bar{X_2}
$$  

📌 This represents the expected **$Y$ value when $X = 0$**.  

---
## **1️⃣.3️⃣ Understanding the Role of Coefficients (β) in Linear Regression**  
The **coefficients** $\beta_1, \beta_2, ..., \beta_n$ determine the **impact of each independent variable $(X)$ on the dependent variable (\(Y\))**.  

The general formula for **multiple linear regression** is:  

$$
Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + ... + \beta_n X_n + \epsilon
$$  

#### Outputs:
🔹 **$Y$ (Dependent Variable):** The outcome being predicted.  

#### Inputs:
🔹 **$X_1, X_2, ..., X_n$ (Independent Variables):** Features influencing $Y$.  

🔹 **$\beta_0$ (Intercept):** The value of $Y$ when all $X_i$ are zero.  

🔹 **$\beta_1, \beta_2, ..., \beta_n$ (Coefficients):** Change in $Y$ for a **one-unit increase** in $X$.  

🔹 **$\epsilon$ (Error term):** Variation in $Y$ not explained by $X$.  

---

## **1️⃣.4️⃣ Relationship Between Coefficients and the Dependent Variable**  
The **coefficient’s sign and magnitude** determine the effect on $Y$:  

✔️ **Positive Coefficient ($\beta_i > 0$)** → **Direct Relationship**: As $X_i$ increases, $Y$ increases.  
💡 *Example:* In house price prediction, if $\beta_1 = 5000$ for **square feet**, **each additional square foot raises the price by $5000**.  


❌ **Negative Coefficient ($\beta_i < 0$)** → **Inverse Relationship**: As $X_i$ increases, $Y$ decreases.  
💡 *Example:* In demand forecasting, if $\beta_1 = -10$ for **price**, **increasing price by $1 reduces demand by 10 units**.  


⭕ **Zero Coefficient ($\beta_i = 0$)** → **No Impact**: Changes in $X_i$ do not affect $Y$.  
💡 *Example:* If **“number of windows in a house”** has $\beta = 0$, this feature **does not influence house price prediction**.  

---
## **1️⃣.5️⃣ Relationship Between Variables & Business Use Cases**  
The relationship between independent ($X$) and dependent ($Y$) variables can be:  

✅ **Positive Correlation** → When $X$ increases, $Y$ increases.  
Example: More **ad spending** → **Higher sales**.  

❌ **Negative Correlation** → When $X$ increases, $Y$ decreases.  
Example: Higher **product price** → **Lower demand**.  

➖ **No Correlation** → No impact between $X$ and $Y$.  
Example: Number of **employees in a company** vs. **weather conditions**.  

📌 The strength of these relationships is measured by **correlation coefficient (R²)**.  

---

## **1️⃣.6️⃣ Summary**  
- Linear Regression models the relationship between **dependent ($Y$) and independent variables ($X$)**.  
- **Coefficients ($\beta$) determine the impact of each $X$ on $Y$**.  
- The **slope ($\beta_1$) measures the rate of change**, while the **intercept ($\beta_0$) gives the baseline value**.  
- **Business applications include**: price prediction, demand forecasting, and customer behavior analysis.  

### *📚 Summary of Key Learnings*

|----------------------|---------------------------|
| 🔹 **Simple Linear Regression** | $ Y = \beta_0 + \beta_1 X $ |

## 2️⃣.1️⃣🚗 **Understanding Intercept & Slope in Linear Regression (In Simple Terms)**  

In Linear Regression, we predict a **dependent variable** $( Y )$ using one or more **independent variables** $( X )$.  

The equation is:  
$$
Y = b_0 + b_1 X_1 + b_2 X_2 + \dots + b_n X_n
$$  
where:  
- **$ Y $** → 🎯 Dependent variable (the value we want to predict)  
- **$ b_0 $ (Intercept)** → 🚦 The starting value of $ Y $ when **all $ X $ values are zero**  
- **$ b_1, b_2, ... $ (Slopes or Coefficients)** → 📈 Show how much $ Y $ **changes** when each $ X $ increases by 1 unit  

---  

### 🏎️ **Intuition: Think of It Like Driving a Car**  
- **Intercept $( b_0 )$** → 🚘 Your starting position (where you begin the journey).  
- **Slope $( b_1, b_2 )$** → ⏩ The speed at which you move towards your destination (how fast $ Y $ increases/decreases as $ X $ changes).  
- **$ X $ Variables** → ⛽ The gas pedal (decides how much effect each variable has on reaching the goal).  

💡 **Easy Way to Remember:**  
- The **intercept** is like your **"base salary"** (how much you earn even if you work 0 hours).  
- The **slope** is like your **"hourly wage"** (how much your earnings increase per extra hour worked).  

---  

### 📊 **Relationship with the Dependent Variable**  

### 🔹 **1. Intercept $( b_0 )$**  
- If all $ X $’s are **zero**, $ Y $ starts from **$ b_0 $**.  
- **Example:** A **bank gives a minimum loan of \$5000** to all applicants, even if they have **zero income**. Here, \$5000 is the **intercept**.  

### 🔹 **2. Slope $( b_1, b_2, ... )$**  
- Shows how much $ Y $ **increases or decreases** for every **1-unit** change in $ X $.  
- **If $ b_1 = 3 $, it means for every 1-unit increase in $ X_1 $, $ Y $ increases by 3**.  

---

## 📈 **4️⃣ R-Squared ($ R^2 $)**

The **coefficient of determination** ($ R^2 $) is calculated using the formula:

$$
R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
$$

where:
- 🔹 $ SS_{res} $ = **Residual Sum of Squares (RSS)** = $ \sum (Y - \hat{Y})^2 $
- 🔹 $ SS_{tot} $ = **Total Sum of Squares (TSS)** = $ \sum (Y - \bar{Y})^2 $
- 🔹 $ \bar{Y} $ = Mean of actual $ Y $ values

$$
(Y - \bar{Y})^2 = [158.76, 22.09, 0.09, 43.56, 121.00]

$$
SST = 158.76 + 22.09 + 0.09 + 43.56 + 121.00 = 345.50

The **Total Sum of Squares (TSS)** is **345.50**.



#### 📉 **Step 3: Compute the Sum of Squared Errors (SSE) $ (Y - \bar{Y})^2 $ (Squared Differences):**


From the previous table, the residuals $ (Y - \hat{Y}) $ are:

$$
Y - \hat{Y} = [-0.52, -0.05, -0.16, -0.11, 0.43]

Each residual is squared:

$$

The **Sum of Squared Errors (SSE)** is **0.4955**.

#### 📏 **Step 3: Compute $ R^2 $ (Coefficient of Determination)**

✅ **Final Answer:** $ R^2 = 0.9986 $, meaning the model explains **99.86%** of the variance in loan approvals.

## 🔄 **5️⃣ Adjusted R-Squared ($ R^2_{adj} $)**

$$
R^2_{adj} = 1 - \left( \frac{(1 - R^2) \times (n - 1)}{n - p - 1} \right)

where:
- 🏷 $ n = 5 $ (Number of observations)
- 🏷 $ p = 2 $ (Number of independent variables)

$$
R_{adj}^2 = 1 - \left( \frac{(1 - 0.9986) \times (5 - 1)}{5 - 2 - 1} \right)
$$

$$
R_{adj}^2 = 1 - \frac{0.0014 \times 4}{2}
$$

$$
R_{adj}^2 = 1 - 0.0028 = 0.9972
$$

✅ **Final Answer:** $ R^2_{adj} = 0.9972 $

## 4️⃣.3️⃣ **Matrix Form Method for Linear Regression**  

### 📌 **Formula to Find Coefficients and Intercept**  
The coefficients (**$ b_i $**) and intercept (**$ b_0 $**) are calculated using the **Ordinary Least Squares (OLS)** method.  

#### 📝 **Formula for Coefficients ($ b $)**  
$$  
b = (X^TX)^{-1}X^TY  
$$  
Where:  
- **$ X $** → Matrix of independent variables (features)  
- **$ Y $** → Target variable (dependent variable)  
- **$ X^T $** → Transpose of $ X $  
- **$ (X^TX)^{-1} $** → Inverse of $ X^TX $  

#### 📝 **Formula for Intercept ($ b_0 $)**  
$$  
b_0 = \bar{Y} - b_1\bar{X_1} - b_2\bar{X_2} - ... - b_n\bar{X_n}  
$$  
Where:  
- **$ \bar{Y} $** → Mean of the target variable  
- **$ \bar{X_i} $** → Mean of each independent variable  

---

## 🔢 **Matrix Form Method for Linear Regression**

- **$ \mathbf{X} $** → Design matrix (features + intercept)  
- **$ \mathbf{Y} $** → Target vector  
- **$ \boldsymbol{\beta} $** → Coefficient vector $[b_0, b_1, b_2]^T$

## 4️⃣.4️⃣ **Gradient Descent Method for Linear Regression**  
Gradient Descent is an optimization algorithm used to find the **best parameters** (coefficients and intercept) for linear regression when **matrix inversion is computationally expensive** (especially for large datasets). Instead of solving directly, we iteratively update the parameters until convergence.  

---

## 📌 **1. Linear Regression Formula**  
The linear regression model is:  

$$
Y = b_0 + b_1 X_1 + b_2 X_2 + \dots + b_n X_n
$$

where:  
- 🔹 $ Y $ → **Predicted value** (target).  
- 🔹 $ X_1, X_2, \dots, X_n $ → **Features**.  
- 🔹 $ b_0 $ → **Intercept**.  
- 🔹 $ b_1, b_2, ..., b_n $ → **Coefficients (slopes)**.  

---

## 📌 **2. Cost Function (Loss Function)**  
We use the **Mean Squared Error (MSE)** as the cost function:  

$$
J(b_0, b_1, ..., b_n) = \frac{1}{2m} \sum_{i=1}^{m} (Y_i - \hat{Y}_i)^2
$$

where:  
- 🔹 $ m $ → Number of data points.  
- 🔹 $ Y_i $ → Actual values.  
- 🔹 $ \hat{Y}_i $ → Predicted values.  
- 🔹 The factor **1/2** is added to simplify differentiation.  

---

## 📌 **3. Gradient Descent Algorithm**  
To minimize the cost function, we update each parameter using **partial derivatives**:  

$$
b_j = b_j - \alpha \frac{\partial J}{\partial b_j}
$$

where:  
- 🔹 $ \alpha $ → **Learning rate** (controls step size).  
- 🔹 $ \frac{\partial J}{\partial b_j} $ → **Gradient (slope) of the cost function**.  

For each coefficient $ b_j $:  

$$
b_j = b_j - \alpha \sum_{i=1}^{m} (Y_i - \hat{Y}_i) X_{ij}
$$

For the **intercept** $ b_0 $:  

$$
b_0 = b_0 - \alpha \sum_{i=1}^{m} (Y_i - \hat{Y}_i)
$$

---

## 📌 **4. Step-by-Step Gradient Descent Example**  
Let's use **5 banking loan applicants** again.  

### 📊 **Dataset**  

| Applicant | Income (Lakhs) ($X_1$) | Credit Score ($X_2$) | Loan Approved ($Y$) |
|-----------|-----------------|---------------|---------------|
| 1 | 5.0 | 645 | 19.2 |
| 2 | 6.5 | 720 | 27.1 |
| 3 | 7.0 | 766 | 31.5 |
| 4 | 8.5 | 810 | 38.4 |
| 5 | 9.0 | 880 | 42.8 |

---

### 🔢 **Step 1: Initialize Parameters**  
Let's start with:  
- 🔹 $ b_0 = 0 $ (Intercept)  
- 🔹 $ b_1 = 0 $ (Weight for Income)  
- 🔹 $ b_2 = 0 $ (Weight for Credit Score)  
- 🔹 $ \alpha = 0.0001 $ (Learning rate)  
- 🔹 **Iterations = 10** (for simplicity)  

---

### 🔢 **Step 2: Compute Predictions**  
$$
\hat{Y}_i = b_0 + b_1 X_{1i} + b_2 X_{2i}
$$

Since all $ b $ values are **zero initially**, the first predictions are:  

$$
\hat{Y} = 0
$$

for all applicants.  

---

### 🔢 **Step 3: Compute Gradients**  
Compute **partial derivatives**:  

$$
\frac{\partial J}{\partial b_0} = \sum_{i=1}^{m} (Y_i - \hat{Y}_i)
$$

$$
\frac{\partial J}{\partial b_1} = \sum_{i=1}^{m} (Y_i - \hat{Y}_i) X_{1i}
$$

$$
\frac{\partial J}{\partial b_2} = \sum_{i=1}^{m} (Y_i - \hat{Y}_i) X_{2i}
$$

**Example for first iteration:**  

$$
\frac{\partial J}{\partial b_0} = (20 - 0) + (25 - 0) + (30 - 0) + (35 - 0) + (40 - 0) = 150
$$

$$
\frac{\partial J}{\partial b_1} = (20 - 0) \times 5 + (25 - 0) \times 6.5 + (30 - 0) \times 7 + (35 - 0) \times 8.5 + (40 - 0) \times 9 = 1130
$$

$$
\frac{\partial J}{\partial b_2} = (20 - 0) \times 650 + (25 - 0) \times 700 + (30 - 0) \times 750 + (35 - 0) \times 800 + (40 - 0) \times 850 = 115000
$$

---

### 🔢 **Step 4: Update Parameters**  
Using:  

$$
b_j = b_j - \alpha \frac{\partial J}{\partial b_j}
$$

$$
b_0 = 0 - (0.0001 \times 150) = -0.015
$$

$$
b_1 = 0 - (0.0001 \times 1130) = -0.113
$$

$$
b_2 = 0 - (0.0001 \times 115000) = -11.5
$$

After **1st iteration**:  

$$
b_0 = -0.015, \quad b_1 = -0.113, \quad b_2 = -11.5
$$

---

### 🔢 **Step 5: Repeat Until Convergence**  
After multiple iterations, the coefficients **converge** to values **similar to what we got with the matrix method**:  

$$
b_0 \approx -86.60, \quad b_1 \approx 5.77, \quad b_2 \approx 0.10
$$

---

## 📌 **5. Key Takeaways**  

✅ **Why Gradient Descent?**  
- 🔹 Works well when the dataset is **large** (e.g., millions of rows).  
- 🔹 Avoids **computing the matrix inverse**, which is computationally expensive.  
- 🔹 Supports **stochastic and mini-batch updates**, making it faster.  

✅ **When Not to Use Gradient Descent?**  
- 🔹 If the dataset is **small** (less than 10,000 rows), the **normal equation** is **better**.  

✅ **Hyperparameter Tuning**  
- 🔹 Learning rate $ \alpha $ should be **small enough** to avoid overshooting but **large enough** for faster convergence.

## **5️⃣.1️⃣ Linear Regression Assumptions**
Linear Regression makes certain assumptions that should be met for the model to perform well.

### 📌 **Assumptions: Identification, Solutions & Consequences**  

Linear Regression relies on **five key assumptions**. If these assumptions are violated, predictions become **biased, inaccurate, or misleading**. The table below outlines each assumption, how to check for violations, how to fix them, and the consequences if left unaddressed.


---



---

## **📌 Assumptions of Linear Regression**  

| **🔹 Assumption** | **🔍 How to Identify?** | **🛠 How to Solve?** | **🚨 What Happens if Not Solved?** |  
|------------------------|------------------------------|----------------------------------|-------------------------------------------|  
| **1️⃣ Linearity** - The relationship between independent (\(X\)) and dependent (\(Y\)) variables should be linear. | - **Scatter plot** of \(X\) vs \(Y\) should show a **straight-line** trend.  <br> - **Residual plot** should have **no curved patterns**. | - Apply **log transformation**, **polynomial regression**, or **interaction terms** to capture non-linearity. | - Predictions become **biased** and inaccurate. <br> - Model underfits the data. |  
| **2️⃣ No Multicollinearity** - Independent variables should not be highly correlated. | - Calculate **VIF (Variance Inflation Factor)**. If **VIF > 5**, multicollinearity exists. <br> - Check **correlation matrix** for high correlations (\(r > 0.8\)). | - Remove one of the correlated features. <br> - Use **Principal Component Analysis (PCA)** to reduce dimensionality. <br> - Use **Ridge Regression**, which can handle collinearity. | - Model coefficients become **unstable** and change drastically with small data variations. <br> - Feature importance becomes unreliable. |  
| **3️⃣ Homoscedasticity** - The variance of residuals should be **constant** across all values of \(X\). | - Plot **residuals vs predicted values**. If variance increases/decreases, heteroscedasticity is present. <br> - Perform **Breusch-Pagan test**. | - Apply **log transformation** or **Box-Cox transformation**. <br> - Use **heteroscedasticity-robust regression** models. | - Model gives **inefficient predictions** with inflated error variance. <br> - Confidence intervals become **unreliable**. |  
| **4️⃣ Normality of Residuals** - The residuals should follow a normal distribution. | - Use **Histogram** or **Q-Q plot** to check residual distribution. <br> - Perform **Shapiro-Wilk test** (p < 0.05 → not normal). | - Apply **log transformation** to \(Y\). <br> - Use **bootstrapping methods** to make statistical inferences more robust. | - Hypothesis testing & confidence intervals become **invalid**. <br> - Predictions may be biased, especially in small datasets. |  
| **5️⃣ No Autocorrelation** - Residuals should be independent and not follow a pattern. | - Perform **Durbin-Watson test** (value close to 2 is good; near 0 or 4 indicates autocorrelation). <br> - Plot **residuals over time** for patterns. | - Use **time series models (ARIMA, LSTM)** if data is sequential. <br> - Add **lag variables** to the model. | - Predictions become **over-optimistic**. <br> - Model accuracy is **inflated** artificially, leading to misleading results. |

| **Mean Absolute Error (MAE)** | $ MAE = \frac{1}{n} \sum |Y_i - \hat{Y}_i|
$ | Measures the **average absolute difference** between actual and predicted values. | **Lower is better**. A smaller MAE means the model makes smaller errors on average. | **Simple & intuitive** metric. Less sensitive to outliers than MSE. | **📍 Real Estate**: Predicting **house prices**, where the difference between predicted and actual price matters. |
| **Mean Squared Error (MSE)** | $ MSE = \frac{1}{n} \sum (Y_i - \hat{Y_i})^2 $ | Measures the **average squared error** between actual and predicted values. | **Lower is better**. Penalizes larger errors more than MAE. | **Heavily penalizes large errors**, making it useful when large deviations are costly. | **📈 Stock Market Forecasting**: Helps reduce large errors in **stock price prediction**. |
| **Root Mean Squared Error (RMSE)** | $ RMSE = \sqrt{MSE} $ | **Square root of MSE**, bringing it back to the same unit as the target variable. | **Lower is better**. Easier to interpret since it's in the same unit as the output variable. | **Balances MSE and interpretability**, making it practical for real-world use. | **⚡ Energy Demand Forecasting**: Predicting daily **electricity consumption** with minimal error. |
| **R²(Coefficient of Determination)** | $ R^2 = 1 - \frac{\sum (Y_i - \hat{Y_i})^2}{\sum (Y_i - \bar{Y})^2} $ | Measures **how well** the model explains variance in data (Goodness-of-fit). | Ranges from **0 to 1**. Higher values indicate a better fit. **\( R^2 = 1 \) means perfect fit**. | Helps evaluate **how well independent variables explain the dependent variable**. | **📊 Marketing Analytics**: Measures how well **ad spend predicts sales revenue**. |
| **Adjusted  $R^2$** | $ Adj\ R^2 = 1 - \left(\frac{(1 - R^2)(n - 1)}{n - p - 1}\right)$ | Adjusted version of $ R^2 $ that considers **the number of predictors**. | **More reliable than $ R^2 $ for multiple regression models**. Prevents overfitting. | Penalizes unnecessary variables, ensuring a **robust model**. | **🏦 Credit Scoring**: Evaluating multiple factors (income, credit history) for **loan approvals**. |

---

### 🚀 **Key Insights:**
✔ **MAE** → Simple, **less sensitive to outliers**.  
✔ **MSE** → **Penalizes large errors**, useful when big mistakes are costly.  
✔ **RMSE** → Easier to interpret as it's in the **same unit** as the target variable.  
✔ **$R^2$** → Measures how well the model explains the data, but **doesn't account for extra variables**.  
✔ **$Adjusted \space R^2 $** → Prevents **overfitting**, making it **better for multiple predictors**.

## **Step 3️⃣: Data Preprocessing & Data Split for ML Models**

```python
# Define features & target

X = df.drop(columns=["median_house_value"])
y = df["median_house_value"]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Fit linear regression model
model = LinearRegression()
model.fit(X_train_scaled, y_train)

# Predictions
y_pred = model.predict(X_test_scaled)

# Residuals
residuals = y_test - y_pred
```

## **Step 4️⃣: Model Training and Evaluation**

#### We will train and evaluate several linear regression models.

#### **1. Multiple Linear Regression Model**
#### **2. Lasso Regression Model**
#### **3. Ridge Regression Model**
#### **4. ElasticNet Model**
#### **5. XGBoost Regressor Model**

```python
# 1. Multiple Linear Regression
print('1. Multiple Linear Regression')
# Multiple Linear Regression (using all features)
multiple_model = LinearRegression()
multiple_model.fit(X_train, y_train)
# Predict and evaluate
y_pred_multiple = multiple_model.predict(X_test)
mae_multiple = mean_absolute_error(y_test, y_pred_multiple)
rmse_multiple = np.sqrt(mean_squared_error(y_test, y_pred_multiple))
r2_multiple = round(r2_score(y_test, y_pred_multiple),2)
print(f"Multiple Linear Regression - MAE: {mae_multiple}, RMSE: {rmse_multiple}, R2: {r2_multiple}")
print('')


# 2. Ridge Regression
print('2. Ridge Regression')
# Ridge Regression (with alpha=1.0)
ridge_model = Ridge(alpha=1.2)
ridge_model.fit(X_train, y_train)
# Predict and evaluate
y_pred_ridge = ridge_model.predict(X_test)
mae_ridge = mean_absolute_error(y_test, y_pred_ridge)
rmse_ridge = np.sqrt(mean_squared_error(y_test, y_pred_ridge))
r2_ridge = round(r2_score(y_test, y_pred_ridge),2)
print(f"Ridge Regression - MAE: {mae_ridge}, RMSE: {rmse_ridge}, R2: {r2_ridge}")
print('')



# 3. Lasso Regression
print('3. Lasso Regression')
# Lasso Regression (with alpha=0.1)
lasso_model = Lasso(alpha=0.1)
lasso_model.fit(X_train, y_train)
# Predict and evaluate
y_pred_lasso = lasso_model.predict(X_test)
mae_lasso = mean_absolute_error(y_test, y_pred_lasso)
rmse_lasso = np.sqrt(mean_squared_error(y_test, y_pred_lasso))
r2_lasso = round(r2_score(y_test, y_pred_lasso),2)
print(f"Lasso Regression - MAE: {mae_lasso}, RMSE: {rmse_lasso}, R2: {r2_lasso}")
print('')


# 4. ElasticNet Regression
print('4. ElasticNet Regression')
# ElasticNet Regression (with alpha=0.1 and l1_ratio=0.5)
elasticnet_model = ElasticNet(alpha=0.1, l1_ratio=0.5)
elasticnet_model.fit(X_train, y_train)
# Predict and evaluate
y_pred_elasticnet = elasticnet_model.predict(X_test)
mae_elasticnet = mean_absolute_error(y_test, y_pred_elasticnet)
rmse_elasticnet = np.sqrt(mean_squared_error(y_test, y_pred_elasticnet))
r2_elasticnet = r2_score(y_test, y_pred_elasticnet)
print(f"ElasticNet Regression - MAE: {mae_elasticnet}, RMSE: {rmse_elasticnet}, R2: {r2_elasticnet}")
print(' ')



# 5. Lasso Regression
print('5.Extreme Gradient Boosting Based Ensemble Model - XGBoost Regressor')
xgb = XGBRegressor(n_estimators=700, learning_rate=0.07, max_depth=6)
xgb.fit(X_train, y_train)
# Predict and evaluate
y_pred_xgb = xgb.predict(X_test)
mae_xgb = mean_absolute_error(y_test, y_pred_xgb)
rmse_xgb = np.sqrt(mean_squared_error(y_test, y_pred_xgb))
r2_xgb = (r2_score(y_test, y_pred_xgb))
print(f"XGBoost Regressor - MAE: {mae_xgb}, RMSE: {rmse_xgb}, R2: {r2_xgb}")
print(' ')
```

## **Step 5️⃣: Model Comparisons**

```python
# Create a DataFrame to compare performance metrics
model_comparison = pd.DataFrame({
    'Model': ['Multiple Linear', 'Ridge', 'Lasso', 'ElasticNet','XGBoost Regressor'],
    'MAE': [mae_multiple, mae_ridge, mae_lasso, mae_elasticnet,mae_xgb],
    'RMSE': [rmse_multiple, rmse_ridge, rmse_lasso, rmse_elasticnet,rmse_xgb],
    'R2': [r2_multiple, r2_ridge, r2_lasso, r2_elasticnet,r2_xgb]
})

# Sort by R2
model_comparison.sort_values(by='R2', ascending=False, inplace=True)

model_comparison
```
